### staticmethod
- 첫 번째 인자로 self, cls를 받지 않는다.
- 클래스나 인스턴스와 관계없는 일반적인 유틸리티 함수를 정의할때 사용한다.
- 독립적인 기능이지만 클래스 안에 넣는 것이 구조적으로 더 나을때 사용한다.


| 항목         | classmethod             | staticmethod |
| ---------- | ----------------------- | ------------ |
| 첫 번째 인자    | `cls` (클래스)             | 없음           |
| 클래스 변수 접근  | 가능                      | 불가능          |
| 인스턴스 변수 접근 | 불가능                     | 불가능          |
| 주 용도       | 클래스 상태를 사용하는 메서드        | 유틸리티 함수 정의   |
| 호출 방식      | 클래스.메서드() 또는 인스턴스.메서드() | 동일           |




````python
class B:
  x = 1


  def xx(self, x): # 인스턴스 메소드
    self.x = x


  @classmethod
  def yy(cls, x):  # 클래스 메소드
    cls.x = x


  @staticmethod
  def zz(a, b):  # 스테이틱 메소드
    print(a+b)


  @staticmethod  # 스테이틱 메소드
  def kk():
    print(x)
````

````python

class Book:  # 객체 클레스

    def __init__(self, title=None, author=None, page=None):
        self.__title = title
        self.__author = author
        self.page = page

    def getTitle(self):
        return self.__title

    def setTitle(self, title):
        self.title = title

    def getAuthor(self):
        return self.__author

    def setAuthor(self, author):
        self.author = author

    def getPage(self):
        return self.__page

    def setPage(self, page):
        self.page = page

    def toString(self):
        return f"{self.title}의 저자는 {self.author}이고, {self.page}페이지입니다."
````

````python
class Dog:
    count = 0

    def __init__(self, name):
        self.name = name
        Dog.count = Dog.count + 1

    @classmethod   
    def get_count(cls):
        return cls.count

    @staticmethod
    def is_valid_name(name):
        # name의 클래스가 문자열이고 길이는 0보다 크면 된다.
        return isinstance(name, str) and len(name) > 0

````
