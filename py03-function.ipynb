
global은 내부와 외부의 객체를 동기화(sync)시켜준다.  
global은 사용하기는 편하지만 언제 바뀌었는지 흐름을 파악하기 힘들기 때문에 남용은 하지 않는다.


```python
# 어쩔 수 없이(남용X) 함수 안에서 함수 밖 변수를 사용해야할 때
c = 3
def proc() :
    global c # 외부 변수 값을 변경하기 전에 global로 선언을 먼저 해준 후 변경
    c = c + 1
    print (c)
```

```python
proc()
```

외부 함수의 자원을 내부 함수에서 참조는 가능하지만 변경은 불가.
변경을 할 때는 nonlocal을 사용한다.


```python
k = 1
def outer():
    m = 1
    def inner(): # 함수 안에 함수 적용 가능
        global k #외부 자원을 함수 안에서 사용할 때
        k = k + 1
        return k, m
    return inner()
```

```python
print(outer())
```

```python
k = 1
def outer():
    m = 1
    def inner(): 
        global k 
        k = k + 1
        m = m + 1
        return k, m
    return inner()
```

```python
print(outer()))
```

```python
k = 1
def outer():
    m = 1
    def inner(): 
        global k 
        k = k + 1
        nonlocal m
        m = m + 1
        return k, m
    return inner()
```

```python
print(outer())
```
